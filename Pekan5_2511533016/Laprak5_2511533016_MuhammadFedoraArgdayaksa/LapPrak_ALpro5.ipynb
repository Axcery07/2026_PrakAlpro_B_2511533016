## 1.1 Latar Belakang
Dalam pemrograman komputer, banyak pekerjaan yang harus dilakukan berulang kali, seperti menampilkan teks yang sama, memproses setiap item dalam daftar belanja, atau menghitung nilai setiap mahasiswa dalam satu kelas. Mengerjakannya secara manual tidak efisien. Karena itu, bahasa pemrograman menyediakan perulangan (*looping*) yang berfungsi menyuruh komputer mengerjakan instruksi secara berulang-ulang.

Salah satu bentuk perulangan pada Python adalah perulangan *for*. Perulangan ini disebut *counted loop* karena biasanya dipakai ketika banyaknya pengulangan sudah diketahui. Perulangan *for* dirancang untuk melakukan iterasi pada urutan data (*sequence*) seperti list, string, atau rentang angka dari fungsi range(). Penulisannya harus memperhatikan titik dua dan indentasi, karena kode di luar blok tidak ikut diulang.

Perulangan *for* juga dapat diletakkan di dalam perulangan *for* lainnya, yang disebut *nested for* atau perulangan bersarang. Pada struktur ini diperlukan dua variabel penghitung, satu untuk perulangan luar (*outer loop*) dan satu untuk perulangan dalam (*inner loop*). Contoh yang umum adalah pembuatan pola bertingkat, dengan perulangan luar mengatur baris dan perulangan dalam mengatur kolom.

Berdasarkan uraian tersebut, praktikum ini disusun untuk memperkenalkan konsep perulangan *for* dan *nested for* pada Python sebagai fondasi dalam membangun program yang efisien dan mampu mengolah data dalam jumlah banyak.

## 2.1 Dasar Teori
### 2.1.1Perulangan for
Perulangan for merupakan salah satu jenis loop dalam Python yang bekerja dengan membaca data dari suatu array atau himpunan data. Sitorus menjelaskan bahwa proses perulangan pada for loops dilakukan berdasarkan array atau himpunan data yang telah didefinisikan pada suatu variabel, sehingga berbeda dengan while loop yang terus berjalan selama kondisi bernilai True [1]. Penulisan blok perulangan for wajib disertai indentasi agar Python dapat mengenali batas blok kode yang akan diulang.

### 2.1.2 Hubungan Perulangan dengan Kondisi if, else, dan elif
Perulangan dalam Python memiliki keterkaitan yang erat dengan struktur kondisional. Sitorus menyebutkan bahwa loop merupakan proses yang berjalan ketika menghadapi suatu kondisi, sehingga loop berkaitan langsung dengan if, else, dan elif sebagai proses decision making pada suatu condition statement. Apabila tidak ditemukan pemecahan masalah terhadap suatu condition statement, maka looping akan terjadi secara terus menerus.

### 2.1.3 Perulangan Bersarang (Nested Loop)
Perulangan bersarang atau nested loop merupakan kondisi di mana terdapat perulangan di dalam perulangan lainnya, atau statement di dalam statement lainnya. Sitorus menjelaskan bahwa nested loop secara singkat dapat dipahami sebagai perulangan di dalam perulangan, di mana perulangan dalam akan dieksekusi secara penuh pada setiap satu iterasi perulangan luar. Struktur ini sangat berguna untuk memproses data yang membutuhkan kontrol baris dan kolom secara bersamaan maupun untuk menghasilkan pola-pola berbasis karakter dua dimensi.

### 2.1.4 Kombinasi Perulangan dengan Percabangan
Perulangan for juga sering dikombinasikan dengan struktur percabangan if di dalamnya untuk menyaring atau mengevaluasi nilai pada setiap iterasi secara selektif. Sitorus menyebutkan bahwa proses kerja looping sangat bergantung pada if, else, elif sebagai condition statement dan formulasi yang diterjemahkan pada kode Python itu sendiri. Kombinasi ini memungkinkan program untuk hanya memproses nilai-nilai tertentu yang memenuhi kondisi yang ditetapkan, seperti menyaring bilangan genap atau menambahkan simbol pemisah antar angka secara dinamis.

### 2.1.5 Akumulator dalam Perulangan
Akumulator merupakan variabel yang diinisialisasi sebelum perulangan dimulai dan diperbarui nilainya pada setiap iterasi untuk menampung hasil perhitungan secara bertahap. Sitorus mencontohkan penggunaan variabel iterasi seperti i += 1 di dalam perulangan yang berfungsi memperbarui nilai akumulator pada setiap putaran agar kondisi perulangan dapat dievaluasi dengan benar dan tidak berjalan tanpa batas. Pola akumulator ini banyak diterapkan dalam perulangan untuk menghitung hasil penjumlahan maupun pencacahan nilai secara kumulatif hingga seluruh iterasi selesai dijalankan.

### 2.1.6 Penerapan Nested Loop untuk Pola Karakter
Salah satu penerapan umum dari perulangan bersarang adalah pembentukan pola karakter dua dimensi seperti segitiga, persegi, maupun pola dekoratif lainnya. Sitorus menjelaskan bahwa pada penerapan nested loop, perulangan luar berperan mengontrol jumlah baris sedangkan perulangan dalam mengontrol jumlah karakter yang dicetak pada setiap baris, sehingga kombinasi keduanya menghasilkan pola yang tersusun secara terstruktur. Manipulasi ekspresi matematis pada argumen perulangan memungkinkan jumlah karakter yang dicetak berubah secara dinamis pada setiap baris sehingga menghasilkan berbagai bentuk pola yang diinginkan.

### 2.2.3 Program "perulangan_for3_2511533016.py"

```python
ulang_3016 = int(input("Masukkan jumlah perulangan: "))

jumlah_3016 = 0
for i_3016 in range(1, ulang_3016 + 1):
    print(i_3016, end=" ")
    jumlah_3016 = jumlah_3016 + i_3016

    if i_3016 < ulang_3016:
        print("+", end=" ")
    else:
        print("=", jumlah_3016, end=" ")
print()
print("Jumlah =", jumlah_3016)
```

Bagian ketiga, `for i_3016 in range(1, ulang_3016 + 1):`, berfungsi sebagai pernyataan perulangan `for` yang mengiterasi nilai dari 1 hingga `ulang_3016` menggunakan fungsi `range()` dengan dua argumen, di mana setiap nilai yang dihasilkan disimpan secara bergantian ke dalam variabel `i_3016` pada setiap putaran perulangan.

4. Bagian keempat, `print(i_3016, end=" ")`, berfungsi untuk mencetak nilai `i_3016` pada setiap iterasi ke layar dengan parameter `end=" "` agar setiap angka dicetak pada baris yang sama dengan pemisah spasi.

Perulangan luar ini akan berjalan sebanyak `batas_3016 + 1` kali sehingga menghasilkan jumlah baris dari 0 hingga `batas_3016`.

3. Bagian ketiga, `for j_3016 in range(batas_3016 + 1):`, berfungsi sebagai perulangan dalam yang mengiterasi nilai dari 0 hingga `batas_3016` pada setiap baris yang dihasilkan oleh perulangan luar, di mana setiap nilai disimpan ke dalam variabel `j_3016` yang merepresentasikan indeks kolom pada pola yang sedang dicetak.

4. Bagian keempat, `print(i_3016 + j_3016, end=" ")`, berfungsi untuk mencetak hasil penjumlahan nilai `i_3016` dan `j_3016` pada setiap iterasi perulangan dalam dengan parameter `end=" "` agar setiap angka dicetak pada baris yang sama dengan pemisah spasi. Nilai yang dicetak pada setiap sel merupakan hasil penjumlahan indeks baris dan indeks kolom yang bersesuaian, sehingga menghasilkan pola matriks penjumlahan yang nilainya bertambah secara diagonal.

5. Bagian kelima, `print()`, berfungsi untuk mencetak baris baru setelah seluruh nilai pada satu baris selesai dicetak oleh perulangan dalam, sehingga nilai pada putaran perulangan luar berikutnya akan dicetak mulai dari baris baru. Hasil keseluruhan program ini adalah pola matriks berukuran `(batas_3016 + 1)` x `(batas_3016 + 1)` yang setiap selnya berisi hasil penjumlahan indeks baris dan indeks kolom.

### 2.2.8 Program "nested_for4_2511533016.py"

```python
tinggi_3016 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3016 % 2 != 0:
    print("Tinggi pola harus bilangan genap.")
else :
    a_3016 = tinggi_3016 
    c_3016 = a_3016 
    lebar_3016 = (2 * tinggi_3016) - 2

    for i_3016 in range(1, tinggi_3016 + 1):
        b_3016 = c_3016 + 1

        for j_3016 in range(1, lebar_3016 + 1):

            # Baris atas dan bawah
            if i_3016 == 1 or i_3016 == tinggi_3016:
                if j_3016 == 1 or j_3016 == lebar_3016:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_3016 == 1 or j_3016 == lebar_3016:
                    print("|", end="")
                else :
                    if j_3016 == c_3016:
                        print("<", end="")
                    elif j_3016 == b_3016:
                        print(">", end="")
                    elif j_3016 == (lebar_3016 - c_3016):
                        print("<", end="")
                    elif j_3016 == (lebar_3016 - c_3016 + 1):
                        print(">", end="")
                    elif j_3016 > b_3016 and j_3016 < (lebar_3016 - c_3016):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3016 -= 2

        if a_3016 <= 0:
            c_3016 = (-a_3016) + 2
        else:
            c_3016 = a_3016
```

1. Bagian pertama, `tinggi_3016 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))`, berfungsi untuk menerima masukan tinggi pola dari pengguna melalui fungsi `input()`, kemudian mengonversinya menjadi tipe data integer menggunakan fungsi `int()`, dan menyimpannya ke dalam variabel `tinggi_3016`.

2. Bagian kedua, `if tinggi_3016 % 2 != 0:`, berfungsi sebagai percabangan `if-else` untuk memvalidasi apakah nilai yang dimasukkan merupakan bilangan genap menggunakan operator modulus `%`. Apabila sisa bagi terhadap 2 tidak sama dengan 0, program akan mencetak pesan peringatan dan tidak menjalankan blok `else`, sedangkan apabila nilai merupakan bilangan genap maka seluruh blok `else` akan dieksekusi.

3. Bagian ketiga, inisialisasi variabel `a_3016 = tinggi_3016`, `c_3016 = a_3016`, dan `lebar_3016 = (2 * tinggi_3016) - 2`, berfungsi untuk menyiapkan variabel-variabel pembantu sebelum perulangan dijalankan. Variabel `a_3016` dan `c_3016` digunakan untuk menghitung posisi karakter `<` dan `>` pada setiap baris, sedangkan `lebar_3016` menentukan lebar total pola yang dicetak.

4. Bagian keempat, `for i_3016 in range(1, tinggi_3016 + 1):`, berfungsi sebagai perulangan luar yang mengiterasi setiap baris dari 1 hingga `tinggi_3016`, di mana variabel `b_3016 = c_3016 + 1` dihitung ulang pada setiap baris untuk menentukan posisi karakter `>` yang berpasangan dengan `<`.

5. Bagian kelima, `for j_3016 in range(1, lebar_3016 + 1):`, berfungsi sebagai perulangan dalam yang mengiterasi setiap kolom dari 1 hingga `lebar_3016` pada setiap baris, di mana karakter yang dicetak ditentukan oleh serangkaian kondisi percabangan berdasarkan posisi `i_3016` dan `j_3016` saat ini.

6. Bagian keenam, kondisi `if i_3016 == 1 or i_3016 == tinggi_3016:`, berfungsi untuk menangani pencetakan baris pertama dan baris terakhir pola. Apabila posisi kolom berada di ujung kiri (`j_3016 == 1`) atau ujung kanan (`j_3016 == lebar_3016`), dicetak karakter `#`, sedangkan posisi kolom lainnya dicetak karakter `=`.

7. Bagian ketujuh, blok `else` pada kondisi baris isi, berfungsi untuk menangani pencetakan baris-baris di antara baris pertama dan terakhir. Karakter `|` dicetak pada kolom paling kiri dan paling kanan sebagai tepi pola, sedangkan karakter `<` dan `>` dicetak pada posisi yang ditentukan oleh variabel `c_3016` dan `b_3016` untuk membentuk pola panah yang bergerak ke dalam dari kedua sisi secara simetris. Posisi di antara kedua pasang panah dicetak dengan karakter `.`, sedangkan posisi lainnya dicetak dengan karakter spasi.

8. Bagian kedelapan, logika pembaruan variabel `a_3016 -= 2` dan kondisi `if a_3016 <= 0:`, berfungsi untuk memperbarui posisi karakter `<` dan `>` setelah setiap baris selesai dicetak. Nilai `a_3016` dikurangi 2 pada setiap iterasi, dan apabila nilainya sudah mencapai 0 atau negatif maka `c_3016` dihitung menggunakan ekspresi `(-a_3016) + 2` untuk membalik arah pergerakan panah, sehingga menghasilkan pola simetris yang menyempit ke tengah lalu melebar kembali.

### 2.2.9 Program "latihan5_2511533016.py"

```python
tinggi_3016 = int(input("Masukkan tinggi pola: "))

for i_3016 in range(1, tinggi_3016 + 1):
    j_3016 = " " * (tinggi_3016 - i_3016)
    k_3016 = "*" * (2 * i_3016 - 1)

    print(j_3016 + k_3016)
```

1. Bagian pertama, `tinggi_3016 = int(input("Masukkan tinggi pola: "))`, berfungsi untuk menerima masukan tinggi pola dari pengguna melalui fungsi `input()`, kemudian mengonversinya menjadi tipe data integer menggunakan fungsi `int()`, dan menyimpannya ke dalam variabel `tinggi_3016`.

2. Bagian kedua, `for i_3016 in range(1, tinggi_3016 + 1):`, berfungsi sebagai perulangan yang mengiterasi nilai dari 1 hingga `tinggi_3016`, di mana setiap nilai disimpan ke dalam variabel `i_3016` yang merepresentasikan nomor baris pola yang sedang dicetak.

3. Bagian ketiga, `j_3016 = " " * (tinggi_3016 - i_3016)`, berfungsi untuk menghasilkan sejumlah karakter spasi sebagai indentasi di sisi kiri pola pada setiap baris. Jumlah spasi dihitung menggunakan ekspresi `tinggi_3016 - i_3016` sehingga semakin besar nilai `i_3016` maka semakin sedikit spasi yang dihasilkan, yang menyebabkan pola bergeser ke kiri secara bertahap pada setiap baris berikutnya.

4. Bagian keempat, `k_3016 = "*" * (2 * i_3016 - 1)`, berfungsi untuk menghasilkan sejumlah karakter bintang yang akan dicetak pada setiap baris. Jumlah bintang dihitung menggunakan ekspresi `2 * i_3016 - 1` sehingga pada baris pertama dicetak 1 bintang, baris kedua 3 bintang, baris ketiga 5 bintang, dan seterusnya, menghasilkan pertambahan 2 bintang pada setiap baris berikutnya.

5. Bagian kelima, `print(j_3016 + k_3016)`, berfungsi untuk mencetak gabungan antara string spasi `j_3016` dan string bintang `k_3016` ke layar pada setiap iterasi menggunakan operator `+`. Hasil keseluruhan program ini adalah pola segitiga sama kaki yang tersusun dari karakter bintang dengan bagian puncak berada di atas dan melebar ke bawah secara simetris.

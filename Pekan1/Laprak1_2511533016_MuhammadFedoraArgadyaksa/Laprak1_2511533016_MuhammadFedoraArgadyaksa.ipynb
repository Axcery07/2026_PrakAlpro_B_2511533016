## 2.1 Dasar Teori
### 2.1.1 Definisi Python dan alur kerjanya
Python merupakan bahasa pemrograman tingkat tinggi yang dibuat oleh Guido van Rossum. Python banyak digunakan untuk membuat berbagai macam program, seperti program berbasis command line interface (CLI), program GUI untuk aplikasi desktop, aplikasi mobile, aplikasi web, IoT, game, hingga program untuk hacking. Python juga dikenal sebagai bahasa pemrograman yang mudah dipelajari karena struktur sintaksnya rapi dan mudah dipahami, sehingga cocok bagi pemula yang belum pernah menulis kode sebelumnya. Secara umum, alur kerja pembuatan program Python terdiri atas tiga tahap: pembuatan skrip Python menggunakan teks editor, penerjemahan skrip ke dalam kode biner oleh interpreter Python agar dapat dipahami komputer, dan terakhir eksekusi perintah tersebut oleh komputer. 

### 2.1.3 Pengenalan Visual Studio Code (VS Code)
Visual Studio Code merupakan sebuah text editor atau code editor yang banyak digunakan oleh pengembang untuk menulis dan mengelola kode program, termasuk program Python. VS Code dikembangkan oleh Microsoft dan bersifat gratis serta dapat digunakan di berbagai sistem operasi seperti Windows, Linux, dan macOS. Editor ini banyak dipilih baik oleh pemula maupun pengembang berpengalaman karena kemudahan proses instalasi dan penggunaannya, tampilan antarmuka yang sederhana, serta dukungan berbagai fitur seperti syntax highlighting, auto-complete, dan integrasi langsung dengan Git untuk kebutuhan kontrol versi.

### 2.1.4 Pengenalan GitHub
GitHub merupakan sebuah situs web dan layanan berbasis cloud yang membantu pengembang menyimpan, mengelola, serta melacak dan mengontrol perubahan pada kode program mereka. Untuk memahami GitHub, terdapat dua konsep yang saling berkaitan, yaitu version control dan Git. Version control adalah sistem yang membantu pengembang melacak dan mengelola perubahan pada kode proyek perangkat lunak, sedangkan Git adalah sistem version control bersifat open-source yang dibuat oleh Linus Torvalds pada tahun 2005. Dengan memanfaatkan Git, GitHub memungkinkan pengembang untuk bekerja secara aman melalui mekanisme branching (menduplikasi bagian dari kode sumber untuk dikembangkan secara terpisah) dan merging (menggabungkan kembali hasil perubahan ke kode sumber utama), sehingga setiap perubahan dapat dilacak dan dikembalikan apabila diperlukan.

### 2.1.5 Tahapan commit dan push file tugas dan laporan tersebut ke akun github

1. Install Git, Pastikan Git sudah terinstall di komputer. Cek dengan buka terminal (CMD/PowerShell) lalu ketik 'git --version'. Jika belum ada, download dan install dari git-scm.com, lalu restart VS Code.
2. Login akun GitHub di VS Code, Buka VS Code, klik ikon Accounts di pojok kiri bawah (ikon orang), lalu pilih 'Sign in with GitHub'. Browser akan terbuka untuk login ke akun GitHub dan klik 'Authorize Visual Studio Code'.
3. Buka folder project, Buka File > Open Folder, lalu pilih folder yang berisi kode tugas/laporan yang ingin di-upload.
4. Masuk ke Source Control & inisialisasi, Klik ikon Source Control di sidebar kiri (ikon cabang, atau tekan Ctrl+Shift+G). Jika folder belum jadi repository Git, akan muncul tombol 'Initialize Repository' — klik tombol itu.
5. Publish ke GitHub, Klik tombol 'Publish to GitHub'. Pilih apakah repo dibuat Public atau Private, lalu beri nama repository. VS Code otomatis membuat repository baru di akun GitHub kamu dan menghubungkannya — tidak perlu bikin manual di github.com.
6. Edit/tambah kode seperti biasa, Setiap kali ada file baru atau perubahan kode, file tersebut akan muncul di daftar 'Changes' pada panel Source Control dengan tanda huruf (U = baru, M = diubah).
7. Tulis pesan & lakukan commit, Ketik pesan singkat yang menjelaskan perubahan di kotak 'Message', misalnya 'Tambah kode praktikum pekan 3', lalu klik tombol biru '✓ Commit' (atau Ctrl+Enter).
8. Push (Sync Changes) ke GitHub, Setelah commit, klik tombol 'Sync Changes' yang muncul di pojok kiri bawah untuk mengunggah kode ke GitHub. Ulangi langkah edit → commit → sync ini setiap kali ada perubahan baru.

**Penjelasan :**
1. Bagian pertama, """ komentar dua baris atau lebih pertemuan 1 praktikum alpro """, merupakan komentar multi baris yang diapit oleh tiga tanda kutip ganda ("""). Bagian ini tidak dieksekusi oleh Python dan hanya berfungsi sebagai catatan atau keterangan program, misalnya nama praktikum atau pertemuan keberapa program ini dibuat.
2. Bagian kedua, print("Hello World"), berfungsi untuk mencetak teks "Hello World" ke layar sebagai output pertama dari program.
3. Bagian ketiga, print() (tanpa argumen), berfungsi untuk mencetak baris kosong, sehingga output yang dihasilkan tampak memiliki jarak/spasi antar baris.
4. Bagian keempat, print("Program ini menghasilkan") dan print("Empat Baris output"), masing-masing berfungsi untuk mencetak satu baris teks kalimat yang menjelaskan bahwa program ini menghasilkan empat baris output.
5. Bagian kelima, print("/ \\ \\\ /// \\\\\\"), berfungsi untuk mencetak karakter garis miring. Tanda backslash (\) merupakan karakter escape, sehingga \\ akan dicetak sebagai satu karakter backslash tunggal (\). Baris ini menampilkan kombinasi karakter / dan \ sesuai jumlah backslash yang dituliskan pada kode.
6. Bagian keenam, print("Program ini" + " dibuat oleh a/n" + " Muhammad Fedora Argadyaksa dari 2511533016"), berfungsi untuk mencetak sebuah kalimat identitas pembuat program. Operator + di sini digunakan untuk melakukan penggabungan beberapa string terpisah menjadi satu kalimat utuh sebelum ditampilkan ke layar.

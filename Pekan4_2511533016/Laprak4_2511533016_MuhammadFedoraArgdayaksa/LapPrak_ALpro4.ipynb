## 1.1 Latar Belakang
Dalam pemrograman komputer, kemampuan mengendalikan alur eksekusi program merupakan aspek fundamental yang wajib dikuasai oleh setiap pemrogram. Dalam penerapannya, struktur percabangan kerap dikombinasikan dengan operator logika and, or, dan not guna membangun kondisi majemuk yang lebih kompleks. Penulisan blok percabangan juga wajib memperhatikan indentasi agar program dapat dieksekusi tanpa kesalahan.

## 2.1 Dasar Teori
### 2.1.1 Percabangan if Tunggal
Percabangan if tunggal merupakan bentuk percabangan paling dasar dalam Python yang digunakan ketika hanya terdapat satu kondisi yang perlu dievaluasi. Muhardian menjelaskan bahwa percabangan if digunakan saat terdapat satu pilihan keputusan, di mana blok kode di dalamnya hanya akan dieksekusi apabila kondisi yang diberikan bernilai True. Penulisan blok if wajib disertai indentasi berupa tab atau spasi, karena tanpa indentasi yang benar program akan menghasilkan kesalahan saat dieksekusi.

### 2.1.2 Percabangan if-else
Percabangan if-else digunakan apabila terdapat dua kemungkinan jalur eksekusi yang berbeda. Menurut Muhardian, percabangan if-else digunakan saat terdapat dua pilihan keputusan, di mana blok if akan dijalankan ketika kondisi bernilai True dan blok else akan dijalankan ketika kondisi bernilai False. Dengan struktur ini, program selalu memiliki jalur eksekusi yang pasti sehingga tidak ada kemungkinan tidak ada blok yang dijalankan sama sekali

### 2.1.3 Percabangan if-elif-else dengan Operator Logika
Percabangan if-elif-else digunakan untuk menangani lebih dari dua kemungkinan kondisi secara berurutan. Muhardian menyebutkan bahwa kata kunci elif merupakan singkatan dari else if yang berfungsi menambahkan kondisi tambahan apabila kondisi sebelumnya bernilai False. Dalam penerapannya, struktur ini sering dikombinasikan dengan operator logika and, or, dan not untuk membangun kondisi majemuk yang lebih kompleks, sehingga program dapat mengevaluasi lebih dari satu faktor dalam satu ekspresi kondisi sekaligus.

### 2.1.4 Multi-if Terpisah untuk Akumulasi Kondisi
Multi-if terpisah merupakan pendekatan percabangan di mana setiap blok if ditulis secara independen dan tidak saling terhubung satu sama lain. Berbeda dengan elif yang menghentikan evaluasi pada kondisi pertama yang terpenuhi, setiap blok if pada pendekatan ini akan selalu dievaluasi secara berurutan tanpa bergantung pada hasil evaluasi blok sebelumnya. Pendekatan ini sangat berguna ketika dibutuhkan akumulasi nilai dari beberapa kondisi sekaligus, seperti pada perhitungan diskon bertingkat yang menumpuk beberapa potongan harga dalam satu transaksi.

### 2.1.5 Percabangan match-case
Percabangan match-case merupakan fitur modern Python sebagai alternatif untuk menangani banyak kemungkinan nilai dari suatu ekspresi secara lebih terstruktur. Konstruksi ini bekerja dengan mencocokkan nilai dari suatu variabel terhadap serangkaian pola yang didefinisikan pada setiap blok case, dan apabila tidak ada pola yang cocok maka blok case _ akan dieksekusi sebagai kondisi default [1]. Penggunaan match-case menghasilkan kode yang lebih rapi dan mudah dibaca dibandingkan penggunaan banyak elif secara berurutan

### 2.1.6 Percabangan untuk Evaluasi Nilai
Struktur percabangan juga banyak diterapkan dalam program untuk mengevaluasi suatu nilai dan menghasilkan keluaran yang berbeda berdasarkan rentang atau kategori tertentu. Muhardian mencontohkan penerapan if-elif-else untuk menentukan grade nilai akademik, di mana setiap kondisi mewakili satu rentang nilai mulai dari grade A hingga E. Pendekatan ini menunjukkan bahwa if-elif-else merupakan pilihan yang efisien dan terstruktur untuk program yang membutuhkan banyak evaluasi nilai secara bertingkat.

### 2.2.5 Program "multi_if2_2511533016.py"

```python
# Input dari user
total_belanja_3016 = float(input("Masukkan total belanja (Rp) :"))

# Input stsatus member (Mengecek apakah user mengetik 'y' atau 'ya')
input_member_3016 = input("Apakah Anda Member? (y/t) : ").strip().lower()
is_member_3016 = input_member_3016 in ["y", "ya"]

# Input status kode promo (Mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3016 = input("Apakah kode promo valid (y/t) : ").strip().lower()
kode_promo_valid_3016 = input_promo_3016 in ["y", "ya"]

total_diskon_persen_3016 = 0

# Multi if terpisah : setiap kodisi diperiksa secara independen
# Diskon bisa ditumpuk (Akumulasi) jika memenuhi beberap syarat sekaligus

if total_belanja_3016 > 1000000 :
    total_diskon_persen_3016 += 10 #Diskon belanja besar

if is_member_3016 :
    total_diskon_persen_3016 += 5 # Diskon member

if kode_promo_valid_3016 :
    total_diskon_persen_3016 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3016 = total_belanja_3016 * (total_diskon_persen_3016 / 100)
total_bayar_3016 = total_belanja_3016 - nominal_diskon_3016

# output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total diskon : {total_diskon_persen_3016}% (Rp{nominal_diskon_3016:,.0f})")
print(f"Total bayar  : Rp {total_bayar_3016:,.0f}")

print(f"Total diskon yang anda dapatkan : {total_diskon_persen_3016}%")
# Ouput : Total  diskon yang anda dapatkan 305 jika belanja > 1 juta, member, dan kode promo valid
```

5. Bagian kelima, `if total_belanja_3016 > 1000000 :`, berfungsi sebagai blok `if` pertama yang mengevaluasi secara independen apakah total belanja pengguna melebihi Rp1.000.000. Apabila terpenuhi, variabel `total_diskon_persen_3016` akan bertambah sebesar 10% sebagai diskon belanja besar.

6. Bagian keenam, `if is_member_3016 :`, berfungsi sebagai blok `if` kedua yang dievaluasi secara independen untuk memeriksa apakah pengguna adalah member. Apabila nilai `is_member_3016` bernilai `True`, variabel `total_diskon_persen_3016` akan bertambah sebesar 5% sebagai diskon member.

7. Bagian ketujuh, `if kode_promo_valid_3016 :`, berfungsi sebagai blok `if` ketiga yang dievaluasi secara independen untuk memeriksa apakah kode promo yang dimasukkan valid. Apabila nilai `kode_promo_valid_3016` bernilai `True`, variabel `total_diskon_persen_3016` akan bertambah sebesar 15% sebagai diskon voucher. Pendekatan multi-`if` terpisah pada ketiga blok kondisi ini memungkinkan diskon untuk diakumulasikan secara bertumpuk apabila pengguna memenuhi lebih dari satu syarat sekaligus.

8. Bagian kedelapan, `nominal_diskon_3016 = total_belanja_3016 * (total_diskon_persen_3016 / 100)`, berfungsi untuk menghitung nominal diskon dalam satuan rupiah dengan cara mengalikan total belanja dengan persentase diskon yang telah terakumulasi, kemudian hasilnya disimpan ke dalam variabel `nominal_diskon_3016`.

10. Bagian kesepuluh, blok `print()` pada bagian output berfungsi untuk mencetak rincian pembayaran ke layar menggunakan f-string dengan format `:,.0f` yang menghasilkan tampilan angka dengan pemisah ribuan dan tanpa angka desimal, sehingga keluaran program menjadi lebih rapi dan mudah dibaca oleh pengguna.

### 2.2.7 Program "match_case1_2511533016.py"

```python
bulan_3016 = int(input("Masukkan angka bulan(1-12) : "))

match bulan_3016 :
    case 1 :
        print("Januari")
    case 2 :
            print("Februari")
    case 3 :
            print("Maret")
    case 4 :
            print("April")
    case 5 :
            print("Mei")
    case 6 :
            print("Juni")
    case 7 :
            print("Juli")
    case 8 :
            print("Agustus")
    case 9 :
            print("September")
    case 10 :
            print("Oktober")
    case 11 :
            print("November")
    case 12 :
            print("Desember")
    case _:
            print("Angka tidak valid")
```

1. Bagian pertama, `bulan_3016 = int(input("Masukkan angka bulan(1-12) : "))`, berfungsi untuk menerima masukan angka bulan dari pengguna melalui fungsi `input()`, kemudian mengonversinya menjadi tipe data integer menggunakan fungsi `int()`, dan menyimpannya ke dalam variabel `bulan_3016`.

2. Bagian kedua, `match bulan_3016 :`, berfungsi sebagai pernyataan utama struktur `match-case` yang akan mencocokkan nilai dari variabel `bulan_3016` terhadap setiap pola yang didefinisikan pada blok `case` di bawahnya secara berurutan.

3. Bagian ketiga, `case 1` hingga `case 12`, berfungsi sebagai pola pencocokkan untuk setiap kemungkinan nilai bulan dari 1 hingga 12. Apabila nilai `bulan_3016` cocok dengan salah satu `case`, maka blok `print()` di dalamnya akan dieksekusi untuk mencetak nama bulan yang sesuai, yaitu mulai dari Januari untuk `case 1` hingga Desember untuk `case 12`, dan evaluasi `case` selanjutnya tidak akan dijalankan.

4. Bagian keempat, `case _ : print("Angka tidak valid")`, berfungsi sebagai kondisi default yang akan dieksekusi apabila nilai `bulan_3016` tidak cocok dengan seluruh pola `case` yang telah didefinisikan sebelumnya, yaitu ketika pengguna memasukkan angka di luar rentang 1 hingga 12. Simbol `_` pada `match-case` berperan serupa dengan `else` pada struktur `if-elif-else` sebagai penangkap kondisi yang tidak terdefinisi.

# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa struktur percabangan merupakan mekanisme penting dalam pemrograman Python yang memungkinkan program untuk mengambil keputusan secara otomatis berdasarkan kondisi tertentu. Python menyediakan beberapa bentuk percabangan yang masing-masing memiliki fungsi dan kasus penggunaan yang berbeda, yaitu `if` tunggal, `if-else`, `if-elif-else`, multi-`if` terpisah, serta `match-case`.

Percabangan `if` tunggal digunakan ketika hanya terdapat satu kondisi yang perlu dievaluasi, sedangkan `if-else` digunakan apabila terdapat dua kemungkinan jalur eksekusi yang berbeda. Untuk menangani lebih dari dua kemungkinan kondisi secara berurutan, digunakan struktur `if-elif-else` yang dalam penerapannya sering dikombinasikan dengan operator logika `and`, `or`, dan `not` untuk membangun kondisi majemuk yang lebih kompleks. Sementara itu, pendekatan multi-`if` terpisah memungkinkan seluruh blok kondisi dievaluasi secara independen sehingga diskon atau efek dari beberapa kondisi dapat diakumulasikan secara bertumpuk tanpa menghentikan evaluasi pada kondisi pertama yang terpenuhi.

Pada praktik pembuatan program, telah diterapkan seluruh konsep percabangan tersebut, seperti penggunaan `if` tunggal untuk memvalidasi kelogisan masukan, penerapan `if-else` untuk menentukan dua jalur keputusan yang berbeda, penggunaan `if-elif-else` dengan operator logika untuk mengevaluasi kelayakan pengendara wahana berdasarkan umur dan kepemilikan SIM C, penerapan multi-`if` terpisah untuk menghitung akumulasi diskon bertingkat, serta penggunaan `match-case` untuk menentukan nama wahana dan harga satuan berdasarkan nomor paket yang dipilih. Dengan demikian, praktikum ini memberikan pemahaman mendalam mengenai struktur percabangan pada Python sebagai fondasi penting sebelum mempelajari struktur perulangan dan pemrograman berbasis fungsi.

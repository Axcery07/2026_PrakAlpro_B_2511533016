## 2.1 Dasar Teori
### 2.1.1 Pengenalan Tipe Data Boolean
Boolean merupakan salah satu tipe data dasar dalam Python yang hanya memiliki dua kemungkinan nilai, yaitu `True` (benar) dan `False` (salah). Tipe data ini banyak digunakan untuk merepresentasikan hasil dari suatu perbandingan atau kondisi logika, misalnya untuk menentukan apakah suatu syarat terpenuhi atau tidak. Nilai boolean biasanya dihasilkan dari operator perbandingan seperti `==`, `!=`, `>`, `<`, `>=`, dan `<=`, atau dari operator logika seperti `and`, `or`, dan `not`. Penulisan `True` dan `False` di Python wajib diawali huruf kapital, karena penulisan dengan huruf kecil akan menyebabkan program mengalami error.

### 2.1.2 Fungsi Input pada Python
Fungsi `input()` digunakan agar program dapat menerima data secara langsung dari pengguna melalui keyboard, sehingga program menjadi lebih interaktif dan dinamis. Data yang dimasukkan melalui fungsi `input()` secara default akan dibaca dalam bentuk tipe data string, sehingga apabila data tersebut ingin digunakan sebagai angka (integer atau float), pemrogram perlu melakukan konversi tipe data (type casting) terlebih dahulu menggunakan fungsi seperti `int()` atau `float()`.

### 2.1.3 Konsep Konstanta
Konstanta adalah nilai yang bersifat tetap dan tidak berubah selama program dijalankan. Berbeda dengan variabel yang nilainya dapat diubah sewaktu-waktu, konstanta digunakan untuk menyimpan data yang memang seharusnya tidak boleh diubah, misalnya nilai phi, jumlah hari dalam seminggu, atau konfigurasi tetap suatu program. Python sebenarnya tidak memiliki mekanisme konstanta secara eksplisit seperti bahasa pemrograman lain, sehingga penerapannya umumnya menggunakan konvensi penamaan variabel dengan huruf kapital seluruhnya, misalnya `PHI = 3.14`, sebagai penanda bahwa nilai tersebut tidak dimaksudkan untuk diubah.

### 2.1.4 Tipe Data Numerik
Tipe data numerik pada Python digunakan untuk menyimpan data berupa angka, yang terbagi menjadi dua jenis utama, yaitu integer (bilangan bulat) dan float (bilangan pecahan/desimal). Tipe data ini memungkinkan program untuk melakukan berbagai operasi aritmatika dasar, seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus (%), dan perpangkatan (**). Pemahaman terhadap tipe data numerik menjadi dasar penting sebelum mempelajari operasi matematis yang lebih kompleks dalam pemrograman.

### 2.1.5 Tipe Data String
String merupakan tipe data yang digunakan untuk menyimpan dan mengolah data berupa teks atau kumpulan karakter. Penulisan string pada Python diapit oleh tanda kutip tunggal (') atau tanda kutip ganda ("), dan untuk teks yang terdiri atas banyak baris dapat menggunakan tanda kutip tiga (`"""`). String bersifat immutable, artinya nilai yang sudah dideklarasikan tidak dapat diubah secara langsung, namun dapat dimanipulasi melalui penggabungan (concatenation), pengindeksan (indexing), maupun pemotongan (slicing) untuk menghasilkan string baru.

6. Bagian keenam, `umur_3016 = int(input("Masukkan Umur : "))` dan `skor_3016 = float(input("Masukkan Skor Tes Awal : "))`, berfungsi untuk mengambil masukan umur dan skor tes dari pengguna, kemudian melakukan *type casting* menjadi integer (untuk umur) dan float (untuk skor).

7. Bagian ketujuh, `token_3016 = 100 + 3j`, berfungsi untuk mendeklarasikan variabel bertipe data kompleks (complex) sebagai token identifikasi praktikan.

8. Bagian kedelapan, `lulus_3016 = skor_3016 >= BATAS_LULUS`, berfungsi untuk mengevaluasi status kelulusan praktikan dengan membandingkan skor tes terhadap batas kelulusan, menghasilkan nilai boolean (`True` atau `False`).

9. Bagian kesembilan, deretan `print(f"... | Tipe: {type(...)}")` pada bagian "DATA PRAKTIKAN & HASIL PEMERIKSAAN", berfungsi untuk menampilkan seluruh data yang telah diinput dan diproses beserta tipe datanya masing-masing menggunakan fungsi bawaan `type()`.

10. Bagian kesepuluh, `print(f"Batas Minimum Nilai: {BATAS_LULUS}")` dan `print(f"Apakah Dinyatakan Lulus?: {lulus_3016} | Tipe: {type(lulus_3016)}")` pada bagian "STATUS KELULUSAN PRAKTIKUM", berfungsi untuk mencetak batas nilai kelulusan serta status akhir kelulusan praktikan beserta tipe datanya, sebagai penutup tampilan program.

Tipe data boolean digunakan untuk merepresentasikan nilai logika benar (True) atau salah (False), tipe data numerik terdiri atas integer, float, dan complex yang masing-masing memiliki karakteristik berbeda, sedangkan tipe data string digunakan untuk menyimpan dan menampilkan data berupa teks, baik dalam satu baris maupun beberapa baris sekaligus.

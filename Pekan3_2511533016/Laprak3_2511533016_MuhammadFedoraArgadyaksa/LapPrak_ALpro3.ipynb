## 1.1 Latar Belakang
Dalam pemrograman komputer, kemampuan untuk melakukan operasi terhadap data merupakan bagian yang tidak terpisahkan dari proses pembuatan suatu program. Setelah memahami berbagai tipe data dasar, langkah selanjutnya yang perlu dikuasai oleh seorang pemrogram adalah bagaimana data-data tersebut dapat diolah dan dimanipulasi melalui operator, yaitu simbol atau tanda khusus yang digunakan untuk melakukan suatu operasi terhadap satu atau lebih nilai (operand).

Dalam bahasa pemrograman Python, terdapat beberapa jenis operator yang wajib dipahami oleh pemula. Pertama, operator aritmatika digunakan untuk melakukan operasi hitung dasar seperti penjumlahan, pengurangan, perkalian, pembagian, pembagian bulat (floor division), modulus, dan perpangkatan. Kedua, operator assignment (penugasan) digunakan untuk memberikan atau memperbarui nilai suatu variabel, baik secara sederhana menggunakan tanda "=" maupun secara ringkas melalui compound assignment seperti "+=", "-=", "*=", dan sejenisnya.

Selain kedua jenis operator tersebut, Python juga menyediakan operator perbandingan (comparison operator) yang digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean, serta operator logika (logical operator) seperti and, or, dan not yang digunakan untuk menggabungkan atau membalik beberapa kondisi logis dalam satu ekspresi. Di sisi lain, terdapat pula operator bitwise yang digunakan untuk melakukan manipulasi data secara langsung pada level bit, seperti operasi AND, OR, XOR, NOT, serta pergeseran bit ke kiri (left shift) dan ke kanan (right shift), yang umumnya digunakan pada kasus-kasus tertentu seperti pengolahan data biner dan optimasi program.

Selain jenis-jenis operator utama tersebut, Python juga memiliki beberapa operator lain yang tidak kalah penting, seperti operator identity (is dan is not) yang digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di memori, serta operator membership (in dan not in) yang digunakan untuk memeriksa keberadaan suatu nilai dalam struktur data seperti list, tuple, atau string. Pemahaman terhadap urutan pengerjaan operator (operator precedence) juga menjadi hal penting agar hasil dari suatu ekspresi yang melibatkan berbagai jenis operator dapat dihitung dengan tepat dan sesuai dengan yang diharapkan.

## 2.1 Dasar Teori
### 2.1.1 Pengenalan Operator Aritmatika
Operator aritmatika merupakan simbol yang digunakan untuk melakukan operasi hitung matematis terhadap dua atau lebih nilai (operand) dalam bahasa pemrograman Python. Operator ini meliputi penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat atau floor division (`//`), modulus atau sisa bagi (`%`), serta perpangkatan (`**`). Operator pembagian (`/`) pada Python selalu menghasilkan tipe data float meskipun kedua operand yang dibagi bertipe integer, sedangkan operator pembagian bulat (`//`) akan menghasilkan nilai bulat dengan cara membuang bagian desimalnya [1].

### 2.1.2 Operator Assignment
Operator assignment digunakan untuk memberikan atau menyimpan suatu nilai ke dalam variabel. Operator assignment yang paling sederhana adalah tanda sama dengan (`=`), yang berfungsi untuk memberikan nilai di sisi kanan kepada variabel di sisi kiri. Selain itu, Python juga menyediakan compound assignment yang merupakan bentuk penyederhanaan dari operasi assignment yang digabungkan dengan operator aritmatika, seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`. Penggunaan compound assignment memungkinkan penulisan kode menjadi lebih singkat, misalnya penulisan `x += 1` memiliki hasil yang sama dengan `x = x + 1` [1].

### 2.1.3 Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai dan akan selalu menghasilkan tipe data boolean, yaitu `True` atau `False`. Operator ini terdiri atas sama dengan (`==`), tidak sama dengan (`!=`), lebih besar (`>`), lebih kecil (`<`), lebih besar atau sama dengan (`>=`), serta lebih kecil atau sama dengan (`<=`). Operator perbandingan banyak digunakan sebagai dasar dalam pembuatan kondisi pada struktur kontrol seperti percabangan [1].

### 2.1.4 Operator Logika (Logical)
Operator logika digunakan untuk menggabungkan atau membalik nilai dari satu atau lebih ekspresi logika, sehingga menghasilkan nilai boolean. Python menyediakan tiga operator logika utama, yaitu `and` yang bernilai `True` apabila kedua kondisi bernilai benar, `or` yang bernilai `True` apabila salah satu kondisi bernilai benar, dan `not` yang berfungsi membalik nilai kebenaran suatu kondisi. Operator logika sering digunakan bersamaan dengan operator perbandingan untuk membentuk ekspresi kondisi yang lebih kompleks [1].

### 2.1.5 Operator Bitwise
Operator bitwise digunakan untuk melakukan operasi secara langsung pada representasi biner (bit) suatu nilai. Operator ini meliputi AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), pergeseran bit ke kiri atau left shift (`<<`), serta pergeseran bit ke kanan atau right shift (`>>`). Operator bitwise umumnya digunakan pada kasus tertentu yang berkaitan dengan pengolahan data pada level bit, seperti optimasi program atau manipulasi data biner [1].

### 2.1.6 Operator Identity
Selain jenis-jenis operator di atas, Python juga menyediakan operator identity, yaitu `is` dan `is not`, yang digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di dalam memori, bukan sekadar memiliki nilai yang sama. Selain itu, terdapat pula operator membership, yaitu `in` dan `not in`, yang digunakan untuk memeriksa apakah suatu nilai terdapat di dalam struktur data seperti list, tuple, maupun string [1].

### 2.2.2 Program "perbandingan_2511533016.py"

```python
# Buat file dengan nama perbandingan_2511533016.py
# Nama variabel ditambah 4 digit nim terakhit
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam python

angka1_3016 = int(input("Input angka 1: "))
angka2_3016 = int(input("Input angka 2: "))

# Lebih besar dari
hasil_3016 = angka1_3016 > angka2_3016
print("\nOperator Lebih Besar Dari")
print("Hasil = ", hasil_3016)

# Lebih kecil dari
hasil_3016 = angka1_3016 < angka2_3016  
print("\nOperator Lebih Kecil Dari")
print("Hasil = ", hasil_3016)

# Lebih besar dari atau sama dengan
hasil_3016 = angka1_3016 >= angka2_3016
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("Hasil = ", hasil_3016)

# Lebih kecil dari atau sama dengan 
hasil_3016 = angka1_3016 <= angka2_3016
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("Hasil = ", hasil_3016)

# Sama dengan
hasil_3016 = angka1_3016 == angka2_3016
print("\nOperator Sama Dengan")
print("Hasil = ", hasil_3016)

# Tidak sama dengan
hasil_3016 = angka1_3016 != angka2_3016
print("\nOperator Tidak Sama Dengan")
print("Hasil = ", hasil_3016)

# Tmabahkan perbandingan berantau dalam python
hasil_3016 == 0 < angka1_3016 < 100
print("\nOperator Perbandingan Berantai")
print("0 < angka1 < 100: ", hasil_3016)

hasil_3016 == 0 < angka2_3016 < 100
print("0 < angka2 < 100: ", hasil_3016)
```

Namun, karena penulisannya masih menggunakan `==`, baris ini hanya akan mengevaluasi ekspresi tersebut tanpa menyimpan hasilnya ke `hasil_3016`, sehingga nilai yang dicetak pada `print()` berikutnya adalah nilai lama dari `hasil_3016` (hasil dari operasi sebelumnya), bukan hasil perbandingan berantai yang baru.

```python
nama_3016 = input("Nama Pelanggan : ")
status_3016 = input("Masukkan Status Pelanggan (member/nonmember) : ").strip().lower()
total_3016 = int(input("Masukkan Total Belanja : "))
jumlah_3016 = int(input("Masukkan jumlah barang : "))
promo_3016 = input("Masukkan Kode Promo : ").strip().upper()

print("=== SISTEM TRANSAKSI TOKO ===\n")
print("Masukkan Nama Pelanggan : ", nama_3016)
print("Masukkan Stats Pelanggann : ", status_3016)
print("Masukkan Total Belanja : ", total_3016)
print("Masukkan Jumlah Barang : ", jumlah_3016)
print("Masukkan kode Promo : ", promo_3016)

syaratDiskon_3016 = total_3016 >= 200000
syaratJumlah_3016 = jumlah_3016 >= 3

daftarPromo_3016 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]
promoTersedia_3016 = promo_3016 in daftarPromo_3016
promoTidakTersedia_3016 = promo_3016 not in daftarPromo_3016 
statusMember_3016 = status_3016 == "member"
bukanMember_3016 = not statusMember_3016
diskonKhusus_3016 = (syaratJumlah_3016 or syaratDiskon_3016) and statusMember_3016
diskonPromo_3016 = (syaratJumlah_3016 or syaratDiskon_3016) and promoTersedia_3016 

diskon_3016 = total_3016 * 0.15 if diskonKhusus_3016 else 0
rata_3016 = total_3016 / jumlah_3016 

totalPembayaran_3016 = total_3016
totalPembayaran_3016 -= diskon_3016

dataPelanggan_3016 = [nama_3016, status_3016, total_3016]
dataPelanggan2_3016 = dataPelanggan_3016
dataPelanggan3_3016 = dataPelanggan_3016.copy()

identitasSama_3016 = dataPelanggan_3016 is dataPelanggan2_3016
identitasBerbeda_3016 = dataPelanggan_3016 is not dataPelanggan3_3016
nilaiSama_3016 = dataPelanggan_3016 == dataPelanggan3_3016     

bitMember_3016 = 0b0001 if statusMember_3016 else 0b0000
bitBelanja_3016 = 0b0010 if syaratDiskon_3016 else 0b0000
bitJumlah_3016 = 0b0100 if syaratJumlah_3016 else 0b0000
bitPromo_3016 = 0b1000 if promoTersedia_3016 else 0b0000

kodeStatus_3016 = bitMember_3016 | bitBelanja_3016 | bitJumlah_3016 | bitPromo_3016
cekMember_3016 = kodeStatus_3016 & 0b0001
cekPromo_3016 = kodeStatus_3016 & 0b1000
kodeReferensi_3016 = 0b1011
hasilXor_3016 = kodeStatus_3016 ^ kodeReferensi_3016
hasilShift_3016 = kodeStatus_3016 << 1

memberAccess_3016 = bool(kodeStatus_3016 & 0b0001)
promoAccess_3016 = bool(kodeStatus_3016 & 0b1000)
freeShippingAccess_3016 = promo_3016 == "GRATISONGKIR" and promoTersedia_3016

print("\n=== HASIL VALIDASI ===")
print("Belanja >= Rp200000        :", syaratDiskon_3016)
print("Jumlah Barang >= 3         :", syaratJumlah_3016)
print("Status Member              :", statusMember_3016)
print("Kode Promo Tersedia        :", promoTersedia_3016)
print("Mendapatkan Diskon         :", diskonKhusus_3016)
print("Mendapatkan Promo          :", promoTersedia_3016)

print("\n=== HASIL PERHITUNGAN ===")
print("Diskon                     : Rp", int(diskon_3016))
print("Total Pembayaran           : Rp", int(totalPembayaran_3016))
print("Rata-rata Harga Barang     : Rp", round(rata_3016, 2))

print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses             :", format(kodeStatus_3016, "04b"))
print("Member Access              :", memberAccess_3016)
print("Promo Access               :", promoAccess_3016)
print("Free Shipping Access       :", freeShippingAccess_3016)

print("\n=== OPERASI BITWISE ===")
print("Kode Status Transaksi")
print("0001 | 0010 | 0100 | 1000")
print("Kode Biner   :", format(kodeStatus_3016, "04b"))
print("Kode Desimal :", kodeStatus_3016)

print("\nCek Member")
print(format(kodeStatus_3016, "04b"), "& 0001")
print("Hasil Biner   :", format(cekMember_3016, "04b"))
print("Hasil Desimal :", cekMember_3016)

print("\nCek Promo")
print(format(kodeStatus_3016, "04b"), "& 1000")
print("Hasil Biner   :", format(cekPromo_3016, "04b"))
print("Hasil Desimal :", cekPromo_3016)

print("\nPerbandingan Status")
print("Kode Transaksi :", format(kodeStatus_3016, "04b"))
print("Kode Referensi :", format(kodeReferensi_3016, "04b"))
print(format(kodeStatus_3016, "04b"), "^", format(kodeReferensi_3016, "04b"))
print("Hasil Biner   :", format(hasilXor_3016, "04b"))
print("Hasil Desimal :", hasilXor_3016)

print("\nGeser Kiri (Shift)")
print(format(kodeStatus_3016, "04b"), "<< 1")
print("Hasil Biner   :", format(hasilShift_3016, "05b"))
print("Hasil Desimal :", hasilShift_3016)

print("\n=== OPERATOR KEANGGOTAAN & IDENTITAS ===")
print("Promo tersedia (in)        :", promoTersedia_3016)
print("Promo tidak tersedia (not in) :", promoTidakTersedia_3016)
print("data_pelanggan is data_pelanggan2 :", identitasSama_3016)
print("data_pelanggan is not data_pelanggan3 :", identitasBerbeda_3016)
print("data_pelanggan == data_pelanggan3 :", nilaiSama_3016)

print("\n=== SELESAI ===")
```
